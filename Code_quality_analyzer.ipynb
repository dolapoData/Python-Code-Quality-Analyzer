### Excessive Nesting Detection

```python
max_nesting = 0

for line in lines:
    if line.strip() == "":
        continue
    leading_spaces = len(line) - len(line.lstrip())
    nesting_level = leading_spaces // 4   # assuming 4 spaces per indent level
    if nesting_level > max_nesting:
        max_nesting = nesting_level

print("Maximum Nesting Depth:", max_nesting)
```

### Missing Docstrings

### Detecting Error Handling

```python
def analyze_file(file_path):
    try:
        with open(file_path, "r") as file:
            lines = file.readlines()
    except FileNotFoundError:
        print("File not found. Please check the file path.")
        return

    # Basic line counts
    total_lines = len(lines)
    blank_lines = 0
    comment_lines = 0
    for line in lines:
        if line.strip() == "":
            blank_lines += 1
        elif line.strip().startswith("#"):
            comment_lines += 1
    code_lines = total_lines - blank_lines - comment_lines

    # Long lines
    long_lines = 0
    long_line_numbers = []
    for i, line in enumerate(lines):
        if len(line.rstrip()) > 79:
            long_lines += 1
            long_line_numbers.append(i + 1)

    # Function count
    function_count = 0
    for line in lines:
        if line.strip().startswith("def "):
            function_count += 1

    # Unclear variable names
    bad_names = ["x", "y", "z", "a", "b", "c", "tmp", "foo", "bar"]
    unclear_names = []
    for line in lines:
        stripped = line.strip()
        if "=" in stripped:
            variable_name = stripped.split("=")[0].strip()
            if variable_name in bad_names:
                unclear_names.append(variable_name)

    # Duplicate lines
    seen_lines = []
    duplicate_lines = []
    for line in lines:
        clean_line = line.strip()
        if clean_line == "":
            continue
        if clean_line in seen_lines:
            duplicate_lines.append(clean_line)
        else:
            seen_lines.append(clean_line)

    # Function length
    function_lines = {}
    current_function = None
    current_length = 0
    for line in lines:
        stripped = line.strip()
        if stripped.startswith("def "):
            if current_function is not None:
                function_lines[current_function] = current_length
            function_name = stripped.split("(")[0].replace("def ", "")
            current_function = function_name
            current_length = 0
        elif current_function is not None:
            current_length += 1
    if current_function is not None:
        function_lines[current_function] = current_length

    # Max nesting
    max_nesting = 0
    for line in lines:
        if line.strip() == "":
            continue
        leading_spaces = len(line) - len(line.lstrip())
        nesting_level = leading_spaces // 4
        if nesting_level > max_nesting:
            max_nesting = nesting_level

    # Missing docstrings
    missing_docstrings = []
    for i in range(len(lines)):
        stripped = lines[i].strip()
        if stripped.startswith("def "):
            function_name = stripped.split("(")[0].replace("def ", "")
            if i + 1 < len(lines):
                next_line = lines[i + 1].strip()
                if not (next_line.startswith('"""') or next_line.startswith("'''")):
                    missing_docstrings.append(function_name)
            else:
                missing_docstrings.append(function_name)

    # Risky patterns
    risky_patterns = ["eval(", "exec("]
    risky_found = []
    for i in range(len(lines)):
        for pattern in risky_patterns:
            if pattern in lines[i]:
                risky_found.append((i + 1, pattern.replace("(", "")))

    # Error handling
    try_count = 0
    except_count = 0
    for line in lines:
        stripped = line.strip()
        if stripped.startswith("try:"):
            try_count += 1
        elif stripped.startswith("except"):
            except_count += 1

    # ---- Scoring ----
    score = 100
    score -= len(unclear_names) * 3
    score -= long_lines * 2
    long_functions = [name for name, length in function_lines.items() if length > 30]
    score -= len(long_functions) * 5
    if max_nesting > 3:
        score -= (max_nesting - 3) * 6
    score -= len(missing_docstrings) * 2
    score -= len(duplicate_lines) * 3
    score -= len(risky_found) * 5

    if score < 0:
        score = 0
    if score > 100:
        score = 100

    if score >= 90:
        rating = "EXCELLENT"
    elif score >= 80:
        rating = "GOOD"
    elif score >= 70:
        rating = "FAIR"
    elif score >= 60:
        rating = "NEEDS IMPROVEMENT"
    else:
        rating = "POOR"

    # ---- Recommendations ----
    recommendations = []
    if long_functions:
        recommendations.append("Break large functions into smaller functions.")
    if max_nesting > 3:
        recommendations.append("Reduce excessive nesting.")
    if unclear_names:
        recommendations.append("Replace unclear variable names.")
    if missing_docstrings:
        recommendations.append("Add documentation to important functions.")
    if duplicate_lines:
        recommendations.append("Review and remove repeated code.")
    if long_lines:
        recommendations.append("Refactor long lines for readability.")
    if risky_found:
        recommendations.append("Manually review potentially risky patterns.")

    # ---- Print the report ----
    print("=" * 50)
    print(" CODE QUALITY ANALYZER")
    print("=" * 50)
    print("File:", file_path)
    print("-" * 50)
    print("SUMMARY")
    print("-" * 50)
    print("Total Lines:", total_lines)
    print("Code Lines:", code_lines)
    print("Blank Lines:", blank_lines)
    print("Comment Lines:", comment_lines)
    print("Functions:", function_count)
    print("Function Lengths:", function_lines)
    print("Long Lines:", long_lines, "-> at lines", long_line_numbers)
    print("Unclear Variable Names:", unclear_names)
    print("Duplicate Lines:", duplicate_lines)
    print("Max Nesting Depth:", max_nesting)
    print("Missing Docstrings:", missing_docstrings)
    print("Risky Patterns:", risky_found)
    print("Try Blocks:", try_count, "| Except Blocks:", except_count)
    print("-" * 50)
    print("QUALITY SCORE")
    print("-" * 50)
    print("Score:", score, "/100")
    print("Rating:", rating)
    print("-" * 50)
    print("RECOMMENDATIONS")
    print("-" * 50)
    if recommendations:
        for i, rec in enumerate(recommendations, 1):
            print(str(i) + ".", rec)
    else:
        print("No major issues detected.")
    print("=" * 50)
```

### Note on Project-Level Scoring
This folder-level scanner uses a simplified subset of checks (naming, line length, duplication) for a quick comparison table.
The single-file `analyze_file()` report above is the authoritative, complete analysis including nesting, documentation, and severity classification. 
This is a known simplification, documented here per the project's methodology guidance.

```python
import os

def analyze_folder(folder_path):
    results = []
    
    for filename in os.listdir(folder_path):
        if filename.endswith(".py"):
            full_path = folder_path + "/" + filename
            
            with open(full_path, "r") as file:
                lines = file.readlines()
            
            # Reuse the same checks (simplified for the summary table)
            long_lines = 0
            for line in lines:
                if len(line.rstrip()) > 79:
                    long_lines += 1
            
            bad_names = ["x", "y", "z", "a", "b", "c", "tmp", "foo", "bar"]
            unclear_names = 0
            for line in lines:
                stripped = line.strip()
                if "=" in stripped:
                    variable_name = stripped.split("=")[0].strip()
                    if variable_name in bad_names:
                        unclear_names += 1
            
            seen_lines = []
            duplicate_count = 0
            for line in lines:
                clean_line = line.strip()
                if clean_line == "":
                    continue
                if clean_line in seen_lines:
                    duplicate_count += 1
                else:
                    seen_lines.append(clean_line)
            
            score = 100
            score -= unclear_names * 3
            score -= long_lines * 2
            score -= duplicate_count * 3
            if score < 0:
                score = 0
            
            if score >= 90:
                rating = "Excellent"
            elif score >= 80:
                rating = "Good"
            elif score >= 70:
                rating = "Fair"
            elif score >= 60:
                rating = "Needs Improvement"
            else:
                rating = "Poor"
            
            results.append({
                "file": filename,
                "score": score,
                "rating": rating
            })
    
    # Print comparison table
    print("=" * 50)
    print(" PROJECT-LEVEL CODE QUALITY REPORT")
    print("=" * 50)
    print(f"{'File':<20}{'Score':<10}{'Rating'}")
    print("-" * 50)
    for r in results:
        print(f"{r['file']:<20}{r['score']:<10}{r['rating']}")
    print("=" * 50)
    
    # Find best and worst
    if results:
        best = max(results, key=lambda r: r["score"])
        worst = min(results, key=lambda r: r["score"])
        avg_score = sum(r["score"] for r in results) / len(results)
        
        print("Average Score:", round(avg_score, 1))
        print("Best File:", best["file"], "-", best["score"])
        print("Worst File:", worst["file"], "-", worst["score"])
```
